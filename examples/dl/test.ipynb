```python
%matplotlib widget

import torch
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from ipywidgets import FloatSlider, VBox, interact
```

```python
# Funktion, deren Parameter wir verändern wollen
def f(x, y, detection_frequency):
    return y * torch.exp(-1j*2*torch.pi * detection_frequency * x)

# Plot-Funktion für interaktive Darstellung
def plot_function(signal_frequency, detection_frequency):
    x = torch.linspace(-10, 10, 1000, dtype=torch.complex64)
    y = torch.sin(2*torch.pi * signal_frequency * x)

    func = f(x,y,detection_frequency)

    plt.figure(figsize=(8,4))
    plt.plot(func.real,func.imag)
    plt.scatter(torch.mean(func.real),torch.mean(func.imag),c="k",marker = "x")
    plt.xlabel("x")
    plt.ylabel("f(x)")
    plt.grid(True)
    plt.xlim(-1,1)
    plt.ylim(-1,1)
    plt.show()

# Interaktive Slider
interact(
    plot_function,
    signal_frequency = FloatSlider(min=1, max=10, step=0.1, value=1, description="signal_frequency:"),
    detection_frequency = FloatSlider(min=1, max=10, step=0.1, value=1, description="detection_frequency:")
)
```
