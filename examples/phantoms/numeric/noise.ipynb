```python
import torch
import zarr as z
import numpy as np

load_store_dir = 'num_phantom_128_R1'
save_store_dir = 'noise_phantom_128_R1'
```

```python
load_store = z.open(f"/home/jovyan/datasets/{load_store_dir}")

C = torch.from_numpy(np.array(load_store["C"]))
k = torch.from_numpy(np.array(load_store["k"]))
d = torch.from_numpy(np.array(load_store["d"]))
```

```python
for item in d:
    item += item*(torch.rand([1]).item()*0.01)-0.005*item
```

```python
save_store = z.storage.LocalStore(f"/home/jovyan/datasets/{save_store_dir}")
group = z.create_group(
    save_store,
    overwrite=True,
)

group.create_array(
    "C",
    shape=C.shape,
    dtype=np.complex64,
    overwrite=True,
)
group.create_array(
    "k",
    shape=k.shape,
    dtype=np.float32,
    overwrite=True,
)
group.create_array(
    "d",
    shape=d.shape,
    dtype=np.complex64,
    overwrite=True,
)

group["C"] = C.numpy()
group["d"] = d.numpy()
group["k"] = k.numpy()
```
